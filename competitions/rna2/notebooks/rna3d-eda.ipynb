```python
def make_linear_baseline_submission(seq_df, n_structures=5, show_progress=True):
    """
    seq_df: DataFrame with columns [target_id, sequence]
    n_structures: number of predicted structures (default=5)
    
    returns: submission-style DataFrame
    """

    rows = []

    iterator = tqdm(seq_df.iterrows(), total=len(seq_df)) if show_progress else seq_df.iterrows()

    for _, r in iterator:
        tid = r["target_id"]
        seq = r["sequence"]

        for i, base in enumerate(seq):
            resid = i + 1

            row = {
                "ID": f"{tid}_{resid}",
                "resname": base,
                "resid": resid,
            }

            # straight line baseline
            for k in range(1, n_structures + 1):
                row[f"x_{k}"] = resid
                row[f"y_{k}"] = 0.0
                row[f"z_{k}"] = 0.0

            rows.append(row)

    return pd.DataFrame(rows)
```

```python
valid_sub = make_linear_baseline_submission(valid_seq)
```

## Evaluation

```python
#　prepare groud truth
sol = valid_labels.copy()
sub = valid_sub.copy()

# extract target_id
sol['target_id'] = sol['ID'].apply(lambda x: '_'.join(str(x).split('_')[:-1]))
sub['target_id'] = sub['ID'].apply(lambda x: '_'.join(str(x).split('_')[:-1]))
```

```python
results = []
for target_id, group_native in tqdm(sol.groupby('target_id')):
    group_predicted = sub[sub['target_id'] == target_id]

    result = score(group_native, group_predicted, 'ID')

    print(target_id, result)
    results.append(result)

print('Mean score:', sum(results)/len(results), f'(n={len(results)})')
```
